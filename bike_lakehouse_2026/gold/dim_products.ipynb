##Modelling our dim_products tables by joining 1 table from crm and 1 from erp all linked to products details

```python
df = spark.sql(
  """
  SELECT
    row_number() OVER (ORDER BY c.product_id) AS product_sk,
    c.product_id,
    c.sales_product_key AS product_key,
    c.category_id,
    c.product_name,
    c.product_cost,
    c.product_line,
    c.product_start_date,
    c.product_end_date,
    e.category,
    e.sub_category,
    e.maintenance
  FROM silver.crm_products AS c
  JOIN silver.erp_products AS e
    ON c.category_id = e.sub_category_key
  """
)

df.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("gold.dim_products")
```
