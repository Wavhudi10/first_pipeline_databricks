#Modelling our Customers dimension using SQL

##This Combines 3 tables 1 from crm and 2 from erp that all have information 

```python
df = spark.sql(
    """
    SELECT
        row_number() OVER (ORDER BY crm.customer_id) AS customers_sk,
        crm.customer_id,
        crm.first_name,
        crm.last_name,
        crm.marital_status,
        crm.gender,
        crm.created_date,
        erp_d.birth_date,
        erp_l.country
    FROM silver.crm_cutomers AS crm
        JOIN silver.erp_customer_details AS erp_d
    ON crm.customer_key = erp_d.customer_key
        JOIN silver.erp_customer_location AS erp_l
    ON crm.customer_key = erp_l.customer_key
    """
)

df.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("gold.dim_customers")
```
