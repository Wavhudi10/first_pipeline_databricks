## Modelling our Fact table and connecting it to our dimesion tables

```python
df = spark.sql(
    """
    SELECT
        p.product_sk,
        c.customers_sk,
        s.* EXCEPT (product_key, customer_id)
    FROM silver.crm_sales AS s
    JOIN gold.dim_products AS p
        ON s.product_key = p.product_key
    JOIN gold.dim_customers AS c
        ON s.customer_id = c.customer_id
    """
)

df.write.format("delta").option("overwriteSchema", "true").mode("overwrite").saveAsTable("gold.fact_sales")
```
